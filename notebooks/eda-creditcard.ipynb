# EDA — creditcard.csv
EDA on the bank transaction dataset with PCA-transformed features.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')

cc = pd.read_csv('../data/raw/creditcard.csv')
print(cc.shape)
cc.head()
```

## 1. Data Cleaning

```python
print('Missing values:')
print(cc.isnull().sum().sum())

print(f'Duplicates: {cc.duplicated().sum()}')
cc.drop_duplicates(inplace=True)

print('\nData types:')
print(cc.dtypes)
```

## 2. Class Imbalance

```python
class_dist = cc['Class'].value_counts()
print('Class distribution:')
print(class_dist)
print(f'\nFraud rate: {class_dist[1] / len(cc):.4%}')

class_dist.plot(kind='bar', color=['steelblue', 'salmon'])
plt.xticks([0, 1], ['Legit', 'Fraud'], rotation=0)
plt.title('Credit Card Class Distribution')
plt.ylabel('Count')
plt.show()
```

## 3. Univariate Distributions

```python
# Amount and Time distributions
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
cc['Amount'].hist(bins=50, ax=axes[0])
axes[0].set_title('Transaction Amount')
cc['Time'].hist(bins=50, ax=axes[1])
axes[1].set_title('Time (seconds)')
plt.tight_layout()
plt.show()
```

## 4. Bivariate Analysis — PCA Features vs Class

```python
# Correlation of PCA features with Class
v_cols = [c for c in cc.columns if c.startswith('V')]
correlations = cc[v_cols + ['Amount', 'Time']].corrwith(cc['Class']).abs().sort_values(ascending=False)

correlations.head(10).plot(kind='bar', figsize=(10, 4), title='|Correlation| with Class')
plt.ylabel('Absolute Correlation')
plt.tight_layout()
plt.show()

# Amount by class
cc.groupby('Class')['Amount'].plot(kind='kde', legend=True)
plt.title('Amount KDE by Class')
plt.xlabel('Amount')
plt.show()
```

```python
# Save cleaned dataset
cc.to_csv('../data/processed/creditcard_cleaned.csv', index=False)
print('Saved: data/processed/creditcard_cleaned.csv')
```
